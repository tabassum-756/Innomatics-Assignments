```python
#Problem 1
def smart_parking(capacity, logs):
  parked = 0
  for log in logs:
    if log == "IN":
      parked +=1
    elif log == "OUT" and parked > 0:
      parked -=1
  if parked >= capacity:
    return "Full"
  return "Available"
parking_capacity = 50
vehicle_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]
status = smart_parking(parking_capacity, vehicle_logs)
print(status)

#Problem 2
def estimate_delivery_time(distance_km, traffic, weather):
  time_minutes = distance_km * 5
  if traffic == "High":
    time_minutes += 15
  elif traffic == "Medium":
    time_minutes += 7
  if weather == "Rainy":
    time_minutes +=10
  elif weather == "Stormy":
    time_minutes += 20
  print(f"Estimated Delivery Time: {time_minutes} minutes")
estimate_delivery_time(8, "High", "Rainy")

#Problem 3
def seat_occupancy(total_seats, booked_seats):
  occupied = len(booked_seats)
  occupancy_percent = (occupied / total_seats) * 100
  if occupancy_percent >=100:
    status = "Housefull"
  elif occupancy_percent >= 75:
    status = "Almost Full"
  else:
    status = "Available"
  print(f"Occupancy: {int(occupancy_percent)}%")
  print(f"Show Status: {status}")
total_seats = 200
booked_seats = [1]*150
seat_occupancy(total_seats, booked_seats)

#Problem 4
def classify_server_load(cpu_readings):
  average_load = sum(cpu_readings) / len(cpu_readings)
  if average_load < 50:
    status = "Normal"
  elif 50 <= average_load <= 80:
    status = "Warning"
  else:
    status = "Critical"
  print(f"Average CPU Load: {int(average_load)}%")
  print(f"Server Status: {status}")
cpu_readings = [45, 60, 70, 85, 90]
classify_server_load(cpu_readings)

#Problem 5
def monitor_resources(usage):
  overused = [res for res, hours in usage.items() if hours > 8]
  alert = " Yes" if overused else "No"
  print(f"Overused Resources: {', '.join(overused) if overused else 'None'}")
  print(f"Energy Alert: {alert}")
resource_usage = {"Projector": 6, "AC": 9, "Lights": 4}
monitor_resources(resource_usage)

#Problem 6
def manage_event_registration(capacity, registrations):
    confirmed = min(capacity, registrations)
    waitlisted = max(0, registrations - capacity)
    status = "Closed" if registrations >= capacity else "Open"
    print(f"Confirmed Registrations: {confirmed}")
    print(f"Waitlisted Users: {waitlisted}")
    print(f"Registration Status: {status}")
event_capacity = 100
registrations = 105
manage_event_registration(event_capacity, registrations)
```
