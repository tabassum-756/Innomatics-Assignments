```python
#problem 1
def validate_recharge(amount):
  valid_plans = [199,299,399,599]
  if amount<50:
    return False, "Recharge must be atleast 50 rupees"
  if amount not in valid_plans:
    return False, "Invalid plan, choose from the given plan"
  return True, "Recharge is successfull"
while True:
  amt = int(input("enter recharge amount: "))
  valid,msg = validate_recharge(amt)
  print(msg)
  if valid:
    break
  else:
    print("retry again..")

#problem 2
def check_inventory(stock_dict):
  for product, qty in stock_dict.items():
    if qty < 15:
      print(product, "Recorder Alert")
    else:
      print(product, "stock ok")
products = {
    "Soap" : 10,
    "Rice" : 25,
    "Oil" : 12,
    "Sugar" : 18
}
check_inventory(products)

#problem 3
def student_result(marks):
  total=0
  for m in marks:
    total+=m
  avg = total/len(marks)
  if avg >=50:
    return avg, "pass"
  else:
    return avg, "fail"
marks_list=[60,40,55,70]
avg,result = student_result(marks_list)
print("average =", avg)
print("result=", result)

#problem 4
def cab_fare(distance, peak):
  base_fare = 50
  per_km = 12
  fare = base_fare + (per_km*distance)
  if peak:
    fare = fare + (fare*0.25)
  return fare
while True:
  km = float(input("Enter distance in km: "))
  peak = input("peak hour? (yes/no): ").lower()
  total = cab_fare(km, peak)
  print("Total Fare = rupees", total)
  retry = input("retry? (yes/no):").lower()
  if retry != "yes":
    break

#problem 5
def attendance_check(attendance_list):
  present=0
  for day in attendance_list:
    if day == "P":
      present+=1
  percentage = (present / len(attendance_list))*100
  if percentage>=75:
    return percentage, "Eligible"
  else:
    return percentage, "Not Eligible"
attendance = ["P","A","P","P","P","A","P"]
percent,status = attendance_check(attendance)
print("Attendance % =",percent)
print("Status =",status)

  #problem 6
def password_strength(password):
  if len(password)<8:
    return "weak: minimum length should be 8"
  has_digit = False
  has_special = False
  special_chars = "@#$"
  for char in password:
    if char.isdigit():
      has_digit = True
    if char in special_chars:
      has_special = True
  if  not has_digit and  not has_special:
    return "weak: should contain atleast one digit and one special character (@#$)"
  return "strong password"
password = input("enter password:")
print(password_strength(password))
```
