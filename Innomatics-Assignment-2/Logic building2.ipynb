```python
setence="python is easy and python is powerful"
words=setence.split()
unique_words=list(dict.fromkeys(words))
print("Unique count:",len(unique_words))
print("Unique words: {" + ", ".join(f"'{word}'" for word in unique_words) + "}")
```

```python
employees={"Ravi":75000,"Anitha":68000,"Kiran":72000}
highest_salary=float('-inf')
highest_key=float('-inf')
for key,value in employees.items():
    if value>highest_salary:
        highest_salary=value
        highest_key=key
print("Highest Salary:",highest_key,highest_salary)
```

```python
numbers=[45,22,89,10,66]
new_lst=numbers.copy()
max=float('-inf')
min=float('+inf')
for i in numbers:
    if i>max:
        max=i
    if i<min:
        min=i
print("List:",new_lst)
print("Max:",max)
print("Min:",min)
```

```python
prices=[450,1200,899,1500,300]
count=0
for p in prices:
    if p>1000:
        count+=1
print("Products above 1000:",count)
        
```

```python
attendance=["P","P","A","P","P"]
count=0
for i in attendance:
    if i=="P":
        count+=1
att_per=(count/len(attendance))*100
print('Attendance percentage:',att_per)
```
